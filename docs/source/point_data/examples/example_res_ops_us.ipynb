# Access ResOpsUS historical reservoir operations dataset ([res_ops_us](https://hf-hydrodata.readthedocs.io/en/latest/gen_res_ops_us.html))

This notebook provides a walk-through of how to access the reservoir operations dataset, ResOpsUS, described in [Steyaert et. al 2022](https://www.nature.com/articles/s41597-022-01134-7).

Please see the full `res_ops_us` [dataset documentation page](https://hf-hydrodata.readthedocs.io/en/latest/gen_res_ops_us.html) for additional citation references for the details on how this dataset was produced. 

```python
# Import packages
import hf_hydrodata as hf
```

```python
# You need to register on https://hydrogen.princeton.edu/pin 
# and run the following with your registered information
# before you can use the hydrodata utilities
hf.register_api_pin("your_email", "your_pin")
```

## Example: Request reservoir inflow data for sites within a HUC2 region that have at least 1 observation within WY2003

```python
options = {
    "dataset": "res_ops_us", 
    "variable": "reservoir_inflow",
    "temporal_resolution": "daily",
    "aggregation": "mean",
    "huc_id": ["1802"],
    "date_start": "2002-10-01",
    "date_end": "2003-09-30",
    "grid": "conus2",
    "min_num_obs": 1
}
 
inflow_data = hf.get_point_data(options)
inflow_data.head()
```

The above example also works for the variable `reservoir_outflow`.

## Example: Request reservoir evaporation data for a single reservoir ID for WY2003

```python
options = {
    "dataset": "res_ops_us", 
    "variable": "reservoir_evaporation",
    "temporal_resolution": "daily",
    "aggregation": "sum",
    "site_ids": ["ResOpsUS-223"],
    "date_start": "2002-10-01",
    "date_end": "2003-09-30",
}

evaporation_data = hf.get_point_data(options)
evaporation_data.head()
```

The above example also works for the variable `reservoir_storage`.

## Example: Request reservoir water elevation data for a single reservoir ID for WY2003

```python
options = {
    "dataset": "res_ops_us", 
    "variable": "reservoir_water_elevation",
    "temporal_resolution": "daily",
    "aggregation": "-",
    "huc_id": ["1802"],
    "date_start": "2002-10-01",
    "date_end": "2003-09-30",
    "grid": "conus2",
    "min_num_obs": 1
}
 
water_elevation_data = hf.get_point_data(options)
water_elevation_data.head()
```
